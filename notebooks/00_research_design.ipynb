# Research Design

## Critical replication of *A Mean Convection Finite Difference Method for Solving Black Scholes Model for Option Pricing*

**Original author:** An Ning (2023)  
**Original paper:** [arXiv:2308.06808](https://arxiv.org/abs/2308.06808)  
**Replication type:** Independent computational replication and reproducibility audit

## Research question

Can finite-difference and Monte Carlo methods reproduce European option prices under the paper's stated Black-Scholes parameters, and are the reported accuracy claims reproducible from the methodological information provided?

## Planned comparisons

1. Recalculate the analytical Black-Scholes benchmark.
2. Implement explicit finite differences on a 100-by-1,000 grid.
3. Implement Crank-Nicolson on the same grid.
4. Run 100,000-path Monte Carlo simulations with fixed seeds.
5. Audit all four reported tables.
6. Study transparent convection multipliers around the standard value of one.

## Validation rules

- Measure independent numerical error against the standard analytical Black-Scholes formula.
- Use put-call parity and monotonicity tests.
- Record every grid, boundary, and random-seed assumption.
- Retain discrepancies rather than tuning the code to match reported values.

## Reproducibility limitation

The paper does not fully define its flux operators, the function used to construct the tuning parameter, the upper price boundary, code, or random seed. Therefore, this project does not claim an exact reconstruction of MCFDM. It provides a conventional-method replication plus a clearly labelled paper-inspired convection sensitivity analysis.

## Authorship and AI-use statement

AI assistance was used for coding support and debugging. The replication design, parameter checks, interpretation of outputs, documented assumptions, and final conclusions were reviewed by the replication author. All numerical claims are generated by reproducible code and tests in this repository.
